```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

categories = [
    'Sleep & Rest\nHygiene',
    'Academic & Focus\nRestructuring',
    'Home & Hostel\nEnvironment',
    'Social & Mood\nSupport'
]

impact = [4.8, 4.5, 4.7, 4.6]
feasibility = [4.5, 4.2, 3.8, 4.4]

labels = [
    '1. Fixed Sleep & Leisure Times',
    '2. Focus Sprints & Reduced Competition',
    '3. Calm Home Space & Instructor Dialogue',
    '4. Connection & Emotional Expression'
]

fig, ax = plt.subplots(figsize=(8.5, 5.5), dpi=300)

colors = ['#2E8B57', '#1E90FF', '#E65C00', '#C71585']

for i in range(len(categories)):
    ax.scatter(
        feasibility[i],
        impact[i],
        color=colors[i],
        s=400,
        alpha=0.85,
        edgecolors='none',
        zorder=3,
        label=categories[i]
    )

ax.set_xlim(3.2, 5.0)
ax.set_ylim(3.8, 5.2)

ax.axhline(
    4.5,
    color='#cccccc',
    linestyle='--',
    linewidth=1.5,
    zorder=1
)

ax.axvline(
    4.2,
    color='#cccccc',
    linestyle='--',
    linewidth=1.5,
    zorder=1
)

ax.text(
    4.9,
    5,
    'HIGH IMPACT / HIGH FEASIBILITY\n(Quick Wins & Priorities)',
    fontsize=9,
    fontweight='bold',
    color='#2E8B57',
    ha='right'
)

ax.text(
    3.3,
    5,
    'HIGH IMPACT / COMPLEX IMPLEMENTATION\n(Systemic Changes)',
    fontsize=9,
    fontweight='bold',
    color='#E65C00',
    ha='left'
)

ax.set_xlabel(
    'Ease of Implementation / Family Feasibility (1: Difficult -> 5: Easy)',
    fontsize=10,
    labelpad=12,
    color='#444444'
)

ax.set_ylabel(
    'Impact on Distress Reduction (1: Low -> 5: High Impact)',
    fontsize=10,
    labelpad=12,
    color='#444444'
)

ax.grid(
    True,
    linestyle=':',
    alpha=0.6
)

ax.legend(
    loc='lower right',
    fontsize=7.5,
    frameon=False,
    framealpha=0.85,
    ncol=1
)

ax.set_title(
    'Strategic Action Matrix: Evidence-Based Pathways to Distress Reduction',
    fontsize=12.5,
    fontweight='bold',
    pad=25,
    color='#222222'
)

plt.tight_layout()
plt.show()
```
