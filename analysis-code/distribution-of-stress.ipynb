```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
df = pd.read_csv('Stress_Dataset.csv')
print("Columns:", df.columns.tolist())
print("\nShape:", df.shape)
print("\nData info:")
print(df.info())
```

```python
print(df['Which type of stress do you primarily experience?'].value_counts())
print("\nGender counts:", df['Gender'].value_counts())
```

```python
sns.set_theme(style="whitegrid")

stress_counts = df['Which type of stress do you primarily experience?'].value_counts()

labels = ['Eustress\n(Motivating)', 'No Stress\n(Balanced)', 'Distress\n(Overwhelming)']

colors = ['#2E8B57', '#708090', '#E65C00']

fig, ax = plt.subplots(figsize=(8, 5), dpi=300)

bars = ax.bar(labels, stress_counts.values, color=colors, width=0.55, edgecolor='none')

ax.yaxis.grid(True, linestyle='--', alpha=0.5)
ax.xaxis.grid(False)

for bar in bars:
    yval = bar.get_height()
    percentage = (yval / len(df)) * 100
    ax.text(bar.get_x() + bar.get_width()/2.0, yval + 15, f"{yval} ({percentage:.1f}%)", 
            ha='center', va='bottom', fontsize=11, fontweight='bold', color='#333333')

ax.set_title('Distribution of Stress Experience Among Participants', fontsize=14, fontweight='bold', pad=25, color='#222222')
ax.set_ylabel('Number of Individuals', fontsize=11, labelpad=10, color='#444444')
ax.set_ylim(0, 880)

plt.tight_layout()
plt.show()
```
